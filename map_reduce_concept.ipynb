```python
def map_fn(doc_id, text):
    return [(word.lower(), 1) for word in text.split()]

def shuffle(mapped):
    agrupados = {}
    for clave, valor in mapped:
        agrupados.setdefault(clave, []).append(valor)
    return agrupados.items()

def reduce_fn(word, counts):
    return (word, sum(counts))
```

```python
documents = {
    1: "Hola mundo hola",
    2: "Hola MapReduce"
}

mapped = []
for doc_id, text in documents.items():
    mapped.extend(map_fn(doc_id, text))

shuffled = shuffle(mapped)

results = [reduce_fn(word, counts) for word, counts in shuffled]

print(results)
```
