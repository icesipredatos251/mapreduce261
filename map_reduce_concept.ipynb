```python
import re
from collections import defaultdict

def map_fn(doc_id, text):
    words = re.findall(r"\w+", text.lower())
    return [(w, 1) for w in words]

def shuffle(mapped_pairs):
    groups = defaultdict(list)
    for word, count in mapped_pairs:
        groups[word].append(count)
    return list(groups.items())

def reduce_fn(word, counts):
    return (word, sum(counts))

documents = {
    1: "Hola mundo hola",
    2: "Hola MapReduce"
}    

mapped = []
for doc_id, text in documents.items():
    mapped.extend(map_fn(doc_id, text))

shuffled = shuffle(mapped)

results = [reduce_fn(word, counts) for word, counts in shuffled]
print(sorted(results))
```
