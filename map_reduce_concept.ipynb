# MapReduce: 3 operaciones (paso a paso)

Usaremos este ejemplo de documentos:

- Documento 1: `"Hola mundo hola"`
- Documento 2: `"Hola MapReduce"`

Las 3 operaciones de MapReduce son:

1. **Map**: transforma cada palabra en pares `(palabra, 1)`.
2. **Shuffle/Group**: agrupa por clave (palabra), juntando todos sus valores.
3. **Reduce**: suma los valores por clave para obtener el conteo final.

Ejecuta la celda de abajo para ver la salida de cada fase.

```python
from collections import defaultdict

# Datos de ejemplo
documents = {
    1: "Hola mundo hola",
    2: "Hola MapReduce"
}

# 1) MAP
# Convierte cada palabra en (palabra, 1)
def map_fn(doc_id, text):
    words = text.lower().split()
    return [(word, 1) for word in words]

mapped = []
for doc_id, text in documents.items():
    mapped.extend(map_fn(doc_id, text))

print("1) MAP ->", mapped)

# 2) SHUFFLE / GROUP
# Agrupa por palabra: palabra -> [1, 1, 1, ...]
def shuffle_fn(mapped_pairs):
    grouped = defaultdict(list)
    for word, count in mapped_pairs:
        grouped[word].append(count)
    return sorted(grouped.items())

shuffled = shuffle_fn(mapped)
print("2) SHUFFLE/GROUP ->", shuffled)

# 3) REDUCE
# Suma los valores de cada palabra
def reduce_fn(word, counts):
    return (word, sum(counts))

results = [reduce_fn(word, counts) for word, counts in shuffled]
print("3) REDUCE ->", results)
```
