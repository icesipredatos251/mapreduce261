# MapReduce en un solo nodo (Word Count)

**Orden de ejecución:**
1. Entrada (documentos)
2. Map
3. Shuffle
4. Reduce
5. Resultado

```python
from collections import defaultdict

def map_fn(doc_id, text):
    """Map: por cada documento emite (palabra, 1) por cada palabra."""
    words = text.lower().split()
    return [(word, 1) for word in words]

def shuffle(mapped):
    """Shuffle: agrupa por clave (palabra). Retorna [(palabra, [1,1,...]), ...]"""
    groups = defaultdict(list)
    for key, value in mapped:
        groups[key].append(value)
    return list(groups.items())

def reduce_fn(word, counts):
    """Reduce: suma los conteos de una misma palabra."""
    return (word, sum(counts))
```

```python

documents = {
    1: "Hola mundo hola",
    2: "Hola MapReduce"
}

# map
mapped = []
for doc_id, text in documents.items():
    mapped.extend(map_fn(doc_id, text))
print("Después de MAP:", mapped)

#shuffle 
shuffled = shuffle(mapped)
print("Después de SHUFFLE:", shuffled)

# reduce
results = [reduce_fn(word, counts) for word, counts in shuffled]


print("Después de REDUCE:", results)
```
