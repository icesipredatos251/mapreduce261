El algoritmo MapReduce se ejecuta en tres etapas principales: 
La etapa **MAP** procesa cada documento y emite pares (palabra, 1) para cada palabra encontrada, normalizando el texto a minúsculas. 

La etapa **SHUFFLE** agrupa todos los valores asociados a la misma clave (palabra), creando una estructura donde cada palabra tiene una lista de todos sus valores (los 1's). 

La etapa **REDUCE** suma todos los valores de cada lista para obtener el conteo total de cada palabra, produciendo el resultado final con el número de ocurrencias de cada palabra en todos los documentos.

```python
from collections import defaultdict


def map_fn(doc_id, text):
    words = text.lower().split()
    return [(word, 1) for word in words]
def shuffle(mapped):
    grouped = defaultdict(list)
    for word, count in mapped:
        grouped[word].append(count)
    return list(grouped.items())
def reduce_fn(word, counts):
    return word, sum(counts)
```

```python
documents = {
    1: "Hola mundo hola",
    2: "Hola MapReduce"
}

mapped = []
for doc_id, text in documents.items():
    mapped.extend(map_fn(doc_id, text))

print("affter map:", mapped)

shuffled = shuffle(mapped)

print("after shuffle:", shuffled)

results = [reduce_fn(word, counts) for word, counts in shuffled]

print("after reduce:", results)
```
